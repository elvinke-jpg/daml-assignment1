```python
# The number of items in the dataset
print("Number of items:", len(df))
```

```python
from pathlib import Path

# check whether every file listed metadata.csv exists
for file in df["file"]:
    print(file, Path(file).exists())
```

```python
# get all original text files in the data folder
data_files = {str(file) for file in Path("data").glob("*.txt")}

# get all file paths listed in the metadata
metadata_files = df["file"].tolist()

# find files that are missing from the metadata
missing_from_metadata = data_files - set(metadata_files)

# find metadata paths that do not correspond to a file in data folder
not_in_data = set(metadata_files) - data_files

# find file paths that occur more than once in the metadata
duplicate_files = df[df["file"].duplicated()]["file"].tolist()

print("Files missing from metadata:", missing_from_metadata)
print("Metadata paths not found in data:", not_in_data)
print("Duplicate file entries:", duplicate_files)
```

```python
# check whether all IDs are unique
print("Number of unique IDs:", df["id"].nunique())
print("Number of rows:", len(df))
print("Are all IDs unique?", df["id"].is_unique)
```
